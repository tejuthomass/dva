```python
import numpy as np
from scipy import stats

# Given Values
mu0 = 100
x_bar = 103
sigma = 12
n = 100
alpha = 0.05

# (c) Critical Value
z_crit = stats.norm.ppf(1 - alpha / 2)

# (d) Standard Error & Z-statistic
se = sigma / np.sqrt(n)
z = (x_bar - mu0) / se

# (e) Two-tailed p-value
p_val = 2 * stats.norm.sf(abs(z))

# (f) 95% Confidence Interval
moe = z_crit * se
ci_low, ci_high = x_bar - moe, x_bar + moe

# Print Results
print("Critical Value: ±", round(z_crit, 4))
print("Standard Error:", round(se, 4))
print("Z-statistic:", round(z, 4))
print("p-value:", round(p_val, 4))
print("95% CI:", round(ci_low, 4), "to", round(ci_high, 4))

# (g) Decision
if abs(z) > z_crit:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to Reject H0")
```
